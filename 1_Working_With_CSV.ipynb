**3. Opening a csv file from an URL**

```python
import pandas as pd

# Example: CSV file hosted online
url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"

# Read CSV from URL
df = pd.read_csv(url)

df.head()
```

```python
import requests
from io import StringIO

url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

```python
# URL of the CSV file
url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"

# Fetch the file using requests
response = requests.get(url)

# Check if request was successful
if response.status_code == 200:
    # Convert the response text into a file-like object
    csv_data = StringIO(response.text)

    # Read into pandas DataFrame
    df = pd.read_csv(csv_data)

    print(df.head())
else:
    print("Failed to fetch CSV file. Status code:", response.status_code)
```

**14. Convertors**

```python
def rename(name):
  if name=='Royal Challengers Bangalore':
    return "RCB"
  else:
    return name
```

```python
df=pd.read_csv("IPL Matches 2008-2020 (1).csv",converters={'team1':rename})
df.head()
```
