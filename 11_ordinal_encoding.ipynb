```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/feature_engineering/main/customer.csv")
```

```python
df.head()
```

```python
df=df.iloc[:,2:5]
```

```python
df.head()
```

```python
X=df.iloc[:,0:2]
y=df.iloc[:,-1]
```

```python
X_train, X_test, y_train, y_test= train_test_split(X,y, test_size=0.2,random_state=0)
```

```python
oc=OrdinalEncoder(categories=[["Poor","Good","Average"],["School","UG","PG"]])
```

```python
oc.fit(X_train)
```

```python
X_train_trans=oc.transform(X_train)
X_test_trans=oc.transform(X_test)
```

```python
X_train_trans=pd.DataFrame(X_train_trans,columns=X_train.columns)
```

```python
X_train_trans.head(2)
```

```python
X_test_trans=pd.DataFrame(X_test_trans,columns=X_train.columns)
```
