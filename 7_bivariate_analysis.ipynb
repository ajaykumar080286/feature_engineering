```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

```python
flights=sns.load_dataset("flights")
flights.head()
```

```python
titanic=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/feature_engineering/main/train%20(6).csv")
titanic.head()
```

```python
iris=sns.load_dataset("iris")
iris.head()
```

**1. Scatterplot (Numerical - Numerical)**

```python
tips.head(2)
```

```python
sns.scatterplot(x=tips["total_bill"], y=tips["tip"],hue=tips["sex"],style=tips["smoker"],size=tips["size"])
```

**2. Bar Plot (Numerical - Categorical)**

```python
titanic.head(3)
```

```python
sns.barplot(x=titanic["Pclass"], y=titanic["Age"], hue=titanic["Sex"])
#sns.barplot(x=titanic["Pclass"], y=titanic["Fare"])
```

```python
sns.barplot(x=titanic["Pclass"], y=titanic["Fare"], hue=titanic["Sex"])
```

**3. Box Plot (Numerical - Categorical)**

```python
sns.boxplot(x=titanic["Sex"], y=titanic["Age"], hue=titanic["Survived"])
```

**4. Distplot (Numerical - Categorical)**

```python
sns.kdeplot(data=titanic[titanic["Survived"]==0], x="Age", label="Did not survive")
sns.kdeplot(data=titanic[titanic["Survived"]==1], x="Age", label="Survived")
plt.legend()
plt.show()
```

**5. HeatMap (Categorical - Categorical)**

```python
sns.heatmap(pd.crosstab(titanic["Pclass"], titanic['Survived']))
```

```python
(titanic.groupby('Embarked').sum()['Survived']*100).plot(kind="bar")
```

**6. ClusterMap (Categorical - Categorical)**

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

**7. Pairplot**

```python
iris.head()
```

```python
sns.pairplot(iris,hue="species")
plt.legend()
```

**8. Lineplot (Numerical - Numerical)**

```python
flights.head()
```

```python
new=flights.groupby("year").sum(numeric_only=True).reset_index()
```

```python
sns.lineplot(x=new["year"], y=new["passengers"])
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
