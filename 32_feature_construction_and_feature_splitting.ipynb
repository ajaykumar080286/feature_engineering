```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/feature_engineering/main/train.csv")[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.head(1)
```

```python
X=df.iloc[:,0:4]
y=df.iloc[:,-1]
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=10)).round(2)
```

```python
X["family_size"]=X["SibSp"]+X["Parch"]
```

```python
X.head(2)
```

**Feature Construction**

```python
def family_type(num):
  if num==1:
    return 0
  elif num>1 and num>4:
    return 1
  else:
    return 2
```

```python
X["family_type"]=X["family_size"].apply(family_type)
```

```python
X.head(3)
```

```python
X.drop(columns=["Parch","SibSp","family_size"], inplace=True)
```

```python
X.head(3)
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=10)).round(1)
```

**Feature Splitting**

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/feature_engineering/main/train.csv")
```

```python
df.head(3)
```

```python
df["Name"]
```

```python
df["title"]=df["Name"].str.split(', ', expand=True)[1].str.split('.',expand=True)[0]
```

```python
df.head(2)
```

```python
df[['title','Name']]
```

```python
(df.groupby('title')['Survived'].mean()).sort_values(ascending=False)
```

```python
df['Is_Married'] = 0
```

```python
df.head(3)
```

```python
df['Is_Married'].loc[df['title'] == 'Mrs']=1
```
