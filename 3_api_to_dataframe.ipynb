```python
df = pd.DataFrame()
for i in range(1,429):
  resposne=requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}".format(i))
  res=resposne.json()
  res1=res['results']
  temp_df=pd.DataFrame(res1)[["id","backdrop_path","release_date"]]
  df = pd.concat([df, temp_df], ignore_index=True)
```
