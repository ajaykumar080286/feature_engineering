```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/feature_engineering/main/covid_toy.csv")
```

```python
df.head()
```

```python
df["cough"].value_counts()
```

```python
X=df.iloc[:,0:5]
y=df.iloc[:,-1]
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.2, random_state=0)
```

***Simple Imputer-Missing Values***

```python
imp=SimpleImputer()
```

```python
X_train_tf1=imp.fit_transform(X_train[["fever"]])
X_test_tf1=imp.transform(X_test[["fever"]])
```

***OrdinalEncoder***

```python
ode=OrdinalEncoder(categories=[["Mild","Strong"]])
```

```python
X_train_tf2=ode.fit_transform(X_train[["cough"]])
X_test_tf2=ode.transform(X_test[["cough"]])
```

**OneHotEncoding**

```python
ohe=OneHotEncoder(drop="first", sparse_output=False)
```

```python
X_train_tf3=ohe.fit_transform(X_train[["gender","city"]])
X_test_tf3=ohe.transform(X_test[["gender","city"]])
```

```python
X_train_tf4=X_train.drop(columns=["gender","fever", "cough","city"]).values
X_test_tf4=X_test.drop(columns=["gender","fever", "cough","city"]).values
```

```python
X_train_transformed=np.concatenate((X_train_tf4, X_train_tf1, X_train_tf3,X_train_tf2),axis=1)
X_test_transformed=np.concatenate((X_test_tf4, X_test_tf1, X_test_tf3,X_test_tf2),axis=1)
```

**Label Encoding**

```python
lbe=LabelEncoder()
```

```python
y_train=lbe.fit_transform(y_train)
y_test=lbe.transform(y_test)
```

**Column Transformer**

```python
columntrans=ColumnTransformer(
    transformers=[
        ("trf1", SimpleImputer(),["fever"]),
        ("trf2", OrdinalEncoder(categories=[["Mild","Strong"]]),["cough"]),
        ("trf3", OneHotEncoder(drop="first", sparse_output=False), ["gender","city"])

    ]
    ,remainder="passthrough"
    )
```

```python
X_trains_trf=columntrans.fit_transform(X_train)
X_test_trf=columntrans.transform(X_test)
```
