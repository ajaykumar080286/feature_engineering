```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/feature_engineering/main/placement_outlier.csv")
```

```python
plt.figure(figsize=(16,5))
plt.subplot(1,2,1)
sns.kdeplot(df['cgpa'])

plt.subplot(1,2,2)
sns.kdeplot(df['placement_exam_marks'])

plt.show()

#Only if not normal distribution here placement_exam_marks is nor full normal distrbution
#so we can apply IR ,we can not apply for cpga bcz full normal distribuionb ,
#you hjav to apply outlier remove using zscrore and uppoer and lower limit
```

```python
sns.boxplot(df['placement_exam_marks'])
```

```python
Q1=df["placement_exam_marks"].quantile(0.25)
Q3=df["placement_exam_marks"].quantile(0.75)
```

```python
IQR=Q3-Q1
```

```python
IQR
```

```python
upper_limit=Q3+1.5*IQR
lower_limit=Q1-1.5*IQR
```

```python
upper_limit
```

```python
df[df["placement_exam_marks"]>upper_limit]
```

```python
df[df["placement_exam_marks"]<lower_limit]
```

```python
new_df = df[df['placement_exam_marks'] < upper_limit]
```

```python
new_df
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,2,1)
sns.kdeplot(df['placement_exam_marks'])

plt.subplot(2,2,2)
sns.boxplot(df['placement_exam_marks'])

plt.subplot(2,2,3)
sns.kdeplot(new_df['placement_exam_marks'])

plt.subplot(2,2,4)
sns.boxplot(new_df['placement_exam_marks'])

plt.show()
```

```python
new_df_cap = df.copy()

new_df_cap['placement_exam_marks'] = np.where(
    new_df_cap['placement_exam_marks'] > upper_limit,
    upper_limit,
    np.where(
        new_df_cap['placement_exam_marks'] < lower_limit,
        lower_limit,
        new_df_cap['placement_exam_marks']
    )
)
```

```python
new_df_cap
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,2,1)
sns.kdeplot(df['placement_exam_marks'])

plt.subplot(2,2,2)
sns.boxplot(df['placement_exam_marks'])

plt.subplot(2,2,3)
sns.kdeplot(new_df_cap['placement_exam_marks'])

plt.subplot(2,2,4)
sns.boxplot(new_df_cap['placement_exam_marks'])

plt.show()
```
